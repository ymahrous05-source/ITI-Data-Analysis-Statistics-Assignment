```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel(r"D:\Data Analyst\ITI Data Analysis\tips.xlsx")
df.head()


```

```python
# Task 3: Data types of columns
print("Data Types:")
print(df.dtypes)
print("-" * 30)

# Task 4: Customer variables
# sex, smoker

# Task 5: Bill variables
# total_bill, day, time, size

# Task 6: Tip variable
# tip

# Task 7: Factors that might affect tip
# total_bill, size, day, time, sex, smoker
```

```python
# Task 8: Smokers count
print("Smokers count:")
print(df['smoker'].value_counts())
print("-" * 30)

# Task 9: Gender count
print("Gender count:")
print(df['sex'].value_counts())
print("-" * 30)

# Task 10: Visits per day
print("Visits per day:")
print(df['day'].value_counts())
print("-" * 30)

# Task 12: Lunch vs Dinner
print("Time count:")
print(df['time'].value_counts())
print("-" * 30)

# Task 13: Most frequent party size
print("Most frequent party size (mode):", df['size'].mode()[0])

# Tasks 14 & 15: Plots for day and time
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.countplot(data=df, x='day', order=['Thur', 'Fri', 'Sat', 'Sun'])
plt.title('Visits by Day')

plt.subplot(1, 2, 2)
sns.countplot(data=df, x='time')
plt.title('Visits by Time')

plt.tight_layout()
plt.show()
```

```python
# Tasks 16 & 17: Total bill Mean and Median
print("Total Bill Mean:", round(df['total_bill'].mean(), 2))
print("Total Bill Median:", round(df['total_bill'].median(), 2))
print("-" * 30)

# Task 18: Party size mode
print("Party size mode:", df['size'].mode()[0])
print("-" * 30)

# Tasks 19 & 20: Tip Mean and Median
print("Tip Mean:", round(df['tip'].mean(), 2))
print("Tip Median:", round(df['tip'].median(), 2))

# Tasks 21 - 24: Note on distribution
# Since Mean > Median for both total_bill and tip, the distribution is Right-Skewed (Positively Skewed).
```

```python
# Tasks 25 & 26: Total bill Min, Max, Range
bill_min = df['total_bill'].min()
bill_max = df['total_bill'].max()
bill_range = bill_max - bill_min

print(f"Total Bill -> Min: {bill_min}, Max: {bill_max}, Range: {round(bill_range, 2)}")

# Tasks 27 & 28: Tip Min, Max, Range
tip_min = df['tip'].min()
tip_max = df['tip'].max()
tip_range = tip_max - tip_min

print(f"Tip -> Min: {tip_min}, Max: {tip_max}, Range: {round(tip_range, 2)}")
print("-" * 30)

# Tasks 29 - 32: Variance and Standard Deviation
print("Total Bill Variance:", round(df['total_bill'].var(), 2))
print("Total Bill Std:", round(df['total_bill'].std(), 2))
print("Tip Variance:", round(df['tip'].var(), 2))
print("Tip Std:", round(df['tip'].std(), 2))

# Task 33: Total bill has higher variability than tip
# Task 34: Standard Deviation / Variance is most sensitive to extreme values.
# Task 35: Total bill has higher std because bill amounts span a much wider range ($3-$50) compared to tips ($1-$10).
```

```python
# Tasks 36 & 37: Total Bill Quartiles and IQR
q1_b = df['total_bill'].quantile(0.25)
q2_b = df['total_bill'].quantile(0.50)
q3_b = df['total_bill'].quantile(0.75)
iqr_b = q3_b - q1_b

print("Total Bill Quartiles:")
print("Q1:", round(q1_b, 2))
print("Q2:", round(q2_b, 2))
print("Q3:", round(q3_b, 2))
print("IQR:", round(iqr_b, 2))
print("-" * 30)

# Tasks 38 & 39: Tip Quartiles and IQR
q1_t = df['tip'].quantile(0.25)
q2_t = df['tip'].quantile(0.50)
q3_t = df['tip'].quantile(0.75)
iqr_t = q3_t - q1_t

print("Tip Quartiles:")
print("Q1:", round(q1_t, 2))
print("Q2:", round(q2_t, 2))
print("Q3:", round(q3_t, 2))
print("IQR:", round(iqr_t, 2))

# Tasks 42 - 46: Boxplots
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.boxplot(y=df['total_bill'])
plt.title('Boxplot of Total Bill')

plt.subplot(1, 2, 2)
sns.boxplot(y=df['tip'])
plt.title('Boxplot of Tip')

plt.tight_layout()
plt.show()
```

```python
# Tasks 47 - 52: Histograms
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.histplot(df['total_bill'], kde=True)
plt.title('Histogram of Total Bill')

plt.subplot(1, 2, 2)
sns.histplot(df['tip'], kde=True)
plt.title('Histogram of Tip')

plt.tight_layout()
plt.show()

# Tasks 53 - 56: Skewness
bill_skew = df['total_bill'].skew()
tip_skew = df['tip'].skew()

print("Total Bill Skewness:", round(bill_skew, 2))
print("Tip Skewness:", round(tip_skew, 2))

# Note: Both variables are positively skewed (Right-Skewed) because skewness > 0
```

```python
# Tasks 57 & 58: Mean Total Bill and Tip by Day
print("Mean Total Bill and Tip by Day:")
print(df.groupby('day')[['total_bill', 'tip']].mean().round(2))
print("-" * 30)

# Tasks 59 & 60: Mean Total Bill and Tip by Smoker
print("Mean Total Bill and Tip by Smoker:")
print(df.groupby('smoker')[['total_bill', 'tip']].mean().round(2))
print("-" * 30)

# Task 61: Mean Tip by Sex
print("Mean Tip by Sex:")
print(df.groupby('sex')['tip'].mean().round(2))
print("-" * 30)

# Task 62: Mean Tip by Time
print("Mean Tip by Time:")
print(df.groupby('time')['tip'].mean().round(2))
```

```python
# Tasks 66 - 70: Covariance and Correlation
cov_val = df['total_bill'].cov(df['tip'])
corr_val = df['total_bill'].corr(df['tip'])

print("Covariance between Total Bill and Tip:", round(cov_val, 2))
print("Correlation between Total Bill and Tip:", round(corr_val, 2))

# Tasks 71 - 78: Scatter Plot
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='total_bill', y='tip', hue='time')
plt.title('Scatter Plot: Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.show()
```

```python
# Tasks 79 - 84: Linear Regression Parameters
m, c = np.polyfit(df['total_bill'], df['tip'], 1)
r2 = corr_val ** 2

print("Slope (m):", round(m, 4))
print("Intercept (c):", round(c, 4))
print(f"Regression Equation: Tip = {m:.4f} * Total_Bill + ({c:.4f})")
print("R-squared (R2):", round(r2, 4))
print("-" * 30)

# Tasks 85 - 88: Predictions for $30 and $50 bills
tip_30 = m * 30 + c
tip_50 = m * 50 + c

print("Predicted Tip for $30 Bill:", round(tip_30, 2))
print("Predicted Tip for $50 Bill:", round(tip_50, 2))

# Task 89: Regression Plot
plt.figure(figsize=(7, 5))
sns.regplot(data=df, x='total_bill', y='tip', line_kws={'color': 'red'})
plt.title('Linear Regression: Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.show()
```

```python
# Tasks 90 - 95: Basic Sampling & Setup
# Sampling 100 times with sample size n=30
sample_means = []

for i in range(100):
    sample = df['tip'].sample(n=30, random_state=i)
    sample_means.append(sample.mean())

# Tasks 96 - 98: Plotting Sampling Distribution
plt.figure(figsize=(7, 4))
sns.histplot(sample_means, kde=True, color='purple')
plt.title('Sampling Distribution of Mean Tip (100 Samples, n=30)')
plt.xlabel('Sample Mean Tip ($)')
plt.ylabel('Frequency')

plt.show()

# Tasks 99 & 100: Comparing Means
print("Mean of Sample Means:", round(np.mean(sample_means), 2))
print("Original Population Mean Tip:", round(df['tip'].mean(), 2))
```
